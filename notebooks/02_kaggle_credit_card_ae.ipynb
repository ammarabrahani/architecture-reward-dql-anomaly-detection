## Step 1: Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import IsolationForest
from sklearn.metrics import (
    f1_score, precision_score, recall_score,
    classification_report, confusion_matrix,
    roc_auc_score, average_precision_score
)

import tensorflow as tf
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

print('Libraries loaded!')
print('TensorFlow:', tf.__version__)
```

## Step 2: Load & Preprocess Dataset

```python
df = pd.read_csv('creditcard.csv')
df_clean = df.drop(columns=['Time'])
X = df_clean.drop(columns=['Class']).values.astype(np.float32)
y = df_clean['Class'].values

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)
X_train_normal = X_train[y_train == 0]

print(f'Train: {X_train.shape} | Normal: {(y_train==0).sum()} | Fraud: {(y_train==1).sum()}')
print(f'Test:  {X_test.shape}  | Normal: {(y_test==0).sum()}  | Fraud: {(y_test==1).sum()}')
print(f'Anomaly %: {round(y.sum()/len(y)*100, 4)}%')
input_dim = X_train.shape[1]
print(f'Input dimension: {input_dim}')
```

## Step 3: Isolation Forest (Classical Baseline)

```python
print('Training Isolation Forest...')
iso = IsolationForest(n_estimators=100, contamination=0.0017, random_state=42)
iso.fit(X_train)

y_pred_if = iso.predict(X_test)
y_pred_if = np.where(y_pred_if == -1, 1, 0)

f1_if = f1_score(y_test, y_pred_if, average='macro')
pr_if = precision_score(y_test, y_pred_if, average='macro')
re_if = recall_score(y_test, y_pred_if, average='macro')

print(f'\n=== Isolation Forest ===')
print(f'Macro F1:        {f1_if:.4f}')
print(f'Macro Precision: {pr_if:.4f}')
print(f'Macro Recall:    {re_if:.4f}')
print(classification_report(y_test, y_pred_if, target_names=['Normal','Fraud']))
```

## Step 4: Autoencoder Architecture Comparison — Table II

```python
def build_ae(input_dim, arch='deep'):
    inp = Input(shape=(input_dim,))
    if arch == 'simple':
        x = Dense(32, activation='relu')(inp)
    elif arch == 'wide':
        x = Dense(128, activation='relu')(inp)
        x = Dense(128, activation='relu')(x)
    elif arch == 'deep':
        x = Dense(64, activation='relu')(inp)
        x = BatchNormalization()(x); x = Dropout(0.2)(x)
        x = Dense(32, activation='relu')(x)
        x = BatchNormalization()(x); x = Dropout(0.2)(x)
        x = Dense(16, activation='relu')(x)
        x = Dense(32, activation='relu')(x)
        x = BatchNormalization()(x); x = Dropout(0.2)(x)
        x = Dense(64, activation='relu')(x)
    out = Dense(input_dim, activation='linear')(x)
    m = Model(inp, out)
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

ae_results = {}
ae_models  = {}

for arch in ['simple', 'wide', 'deep']:
    print(f'Training {arch} AE...')
    m = build_ae(input_dim, arch)
    m.fit(X_train_normal, X_train_normal, epochs=100, batch_size=256,
          validation_split=0.1,
          callbacks=[EarlyStopping(patience=5, restore_best_weights=True)],
          verbose=0)
    recon  = m.predict(X_test, verbose=0)
    errors = np.mean(np.square(X_test - recon), axis=1)
    thresh = np.percentile(errors, 90)
    y_pred = (errors > thresh).astype(int)
    f1 = f1_score(y_test, y_pred, average='macro')
    pr = precision_score(y_test, y_pred, average='macro')
    re = recall_score(y_test, y_pred, average='macro')
    ae_results[arch] = {'F1': round(f1,4), 'Precision': round(pr,4), 'Recall': round(re,4), 'errors': errors}
    ae_models[arch]  = (m, errors, thresh)
    print(f'  {arch} AE — F1: {f1:.4f} | Precision: {pr:.4f} | Recall: {re:.4f}')

print('\n=== TABLE II — Kaggle AE Results ===')
print(f'{"Architecture":<12} {"F1":>8} {"Precision":>12} {"Recall":>10}')
for arch, v in ae_results.items():
    print(f'{arch.capitalize()+" AE":<12} {v["F1"]:>8} {v["Precision"]:>12} {v["Recall"]:>10}')
print('>>> USE THESE IN TABLE II <<<')
```

## Step 5: DQL Architecture Comparison — Table III

```python
def build_dql(input_dim, arch='deep'):
    if arch == 'simple':
        m = Sequential([Input(shape=(input_dim,)), Dense(32, activation='relu'), Dense(2, activation='linear')])
    elif arch == 'wide':
        m = Sequential([Input(shape=(input_dim,)), Dense(128, activation='relu'), Dense(128, activation='relu'), Dense(2, activation='linear')])
    elif arch == 'deep':
        m = Sequential([Input(shape=(input_dim,)), Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
                        Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
                        Dense(16, activation='relu'), Dense(32, activation='relu'), Dense(2, activation='linear')])
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

def train_dql(X_tr, y_tr, X_te, y_te, arch, n_episodes=20, asymmetric=True):
    idx_fraud  = np.where(y_tr == 1)[0]
    idx_normal = np.where(y_tr == 0)[0]
    idx_n = np.random.choice(idx_normal, size=len(idx_fraud), replace=False)
    sample_idx = np.concatenate([idx_n, idx_fraud])
    np.random.shuffle(sample_idx)
    X_dql = X_tr[sample_idx]
    y_dql = y_tr[sample_idx]
    qm = build_dql(X_dql.shape[1], arch)
    epsilon = 1.0
    f1_eps = []
    for ep in range(n_episodes):
        q_all = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()
        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            if action == y_dql[i]: reward = +1
            elif action == 0 and y_dql[i] == 1: reward = -2 if asymmetric else -1
            else: reward = -1
            targets[i][action] = reward
        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)
        preds = np.argmax(qm.predict(X_te, verbose=0), axis=1)
        f1_eps.append(f1_score(y_te, preds, average='macro'))
    preds = np.argmax(qm.predict(X_te, verbose=0), axis=1)
    return f1_score(y_te, preds, average='macro'), f1_eps, qm

dql_results = {}
dql_models  = {}

np.random.seed(42)
tf.random.set_seed(42)

for arch in ['simple', 'wide', 'deep']:
    print(f'Training DQL {arch}...')
    f1, f1_eps, qm = train_dql(X_train, y_train, X_test, y_test, arch)
    pr = precision_score(y_test, np.argmax(qm.predict(X_test, verbose=0), axis=1), average='macro')
    re = recall_score(y_test, np.argmax(qm.predict(X_test, verbose=0), axis=1), average='macro')
    dql_results[arch] = {'F1': round(f1,4), 'Precision': round(pr,4), 'Recall': round(re,4), 'Convergence': f1_eps}
    dql_models[arch]  = qm
    print(f'  DQL {arch} — F1: {f1:.4f} | Precision: {pr:.4f} | Recall: {re:.4f}')

print('\n=== TABLE III — Kaggle DQL Results ===')
print(f'{"Architecture":<12} {"F1":>8} {"Precision":>12} {"Recall":>10}')
for arch, v in dql_results.items():
    print(f'{"DQL "+arch.capitalize():<12} {v["F1"]:>8} {v["Precision"]:>12} {v["Recall"]:>10}')
print('>>> USE THESE IN TABLE III <<<')
```

## Step 6: DQL Convergence Plot — Fig 4

```python
plt.figure(figsize=(10, 6))
colors = {'simple': '#95a5a6', 'deep': '#e74c3c', 'wide': '#3498db'}

for arch, data in dql_results.items():
    plt.plot(range(1, 21), data['Convergence'],
             marker='o', label=f'DQL {arch.capitalize()} (F1={data["F1"]})',
             color=colors[arch], linewidth=2)

plt.axhline(y=f1_if, color='green', linestyle='--', linewidth=1.5,
            label=f'Isolation Forest ({f1_if:.4f})')
plt.axhline(y=ae_results['deep']['F1'], color='orange', linestyle='--',
            linewidth=1.5, label=f'Deep AE ({ae_results["deep"]["F1"]})')

plt.xlabel('Episode', fontsize=12)
plt.ylabel('Macro F1-Score', fontsize=12)
plt.title('DQL Convergence — Kaggle Credit Card Fraud', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('kaggle_dql_convergence.png', dpi=150)
plt.show()
print('Plot saved: kaggle_dql_convergence.png')
```

## Step 7: 5-Seed Statistical Validation — Table VI

```python
seeds = [42, 0, 7, 123, 2024]
f1_seeds, pr_seeds, re_seeds = [], [], []

print('Running DQL Deep with 5 seeds...')
for seed in seeds:
    np.random.seed(seed)
    tf.random.set_seed(seed)
    f1, _, qm = train_dql(X_train, y_train, X_test, y_test, 'deep')
    preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
    pr = precision_score(y_test, preds, average='macro')
    re = recall_score(y_test, preds, average='macro')
    f1_seeds.append(f1); pr_seeds.append(pr); re_seeds.append(re)
    print(f'  Seed {seed:4d} → F1: {f1:.4f} | Precision: {pr:.4f} | Recall: {re:.4f}')

print(f'\n=== TABLE VI — Kaggle 5-Seed Summary ===')
print(f'Macro F1  : {np.mean(f1_seeds):.4f} ± {np.std(f1_seeds):.4f}')
print(f'Precision : {np.mean(pr_seeds):.4f} ± {np.std(pr_seeds):.4f}')
print(f'Recall    : {np.mean(re_seeds):.4f} ± {np.std(re_seeds):.4f}')
print('>>> USE THESE IN TABLE VI <<<')
```

## Step 8: Symmetric vs Asymmetric Reward — Table IV

```python
asym_f1s, sym_f1s = [], []

print('Comparing reward schemes (5 seeds)...')
for seed in seeds:
    np.random.seed(seed); tf.random.set_seed(seed)
    a, _, _ = train_dql(X_train, y_train, X_test, y_test, 'deep', asymmetric=True)
    np.random.seed(seed); tf.random.set_seed(seed)
    s, _, _ = train_dql(X_train, y_train, X_test, y_test, 'deep', asymmetric=False)
    asym_f1s.append(a); sym_f1s.append(s)
    print(f'  Seed {seed} → Asymmetric: {a:.4f} | Symmetric: {s:.4f}')

asym_mean = np.mean(asym_f1s)
sym_mean  = np.mean(sym_f1s)
improvement = ((asym_mean - sym_mean) / sym_mean) * 100

print(f'\n=== TABLE IV — Kaggle Reward Comparison ===')
print(f'Symmetric  : {sym_mean:.4f} ± {np.std(sym_f1s):.4f}')
print(f'Asymmetric : {asym_mean:.4f} ± {np.std(asym_f1s):.4f}')
print(f'Improvement: {improvement:.2f}%')
print('>>> USE THESE IN TABLE IV <<<')
```

## Step 9: Extended Metrics (FPR, ROC-AUC, PR-AUC) — Table V

```python
def get_ext_metrics(y_true, y_pred, y_score, name):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    f1  = f1_score(y_true, y_pred, average='macro')
    pr  = precision_score(y_true, y_pred, average='macro')
    re  = recall_score(y_true, y_pred, average='macro')
    fpr = fp / (fp + tn)
    roc = roc_auc_score(y_true, y_score)
    prc = average_precision_score(y_true, y_score)
    print(f'{name}: F1={f1:.4f} | Prec={pr:.4f} | Rec={re:.4f} | FPR={fpr:.4f} | ROC={roc:.4f} | PR={prc:.4f}')
    return {'Model': name, 'F1': round(f1,4), 'Prec': round(pr,4),
            'Recall': round(re,4), 'FPR': round(fpr,4),
            'ROC-AUC': round(roc,4), 'PR-AUC': round(prc,4)}

print('=== TABLE V — Kaggle Extended Metrics ===')
ext_results = []

# IF
if_scores = -iso.score_samples(X_test)
ext_results.append(get_ext_metrics(y_test, y_pred_if, if_scores, 'Isolation Forest'))

# Deep AE
_, ae_errors, _ = ae_models['deep']
thresh_ae = np.percentile(ae_errors, 90)
y_pred_ae = (ae_errors > thresh_ae).astype(int)
ext_results.append(get_ext_metrics(y_test, y_pred_ae, ae_errors, 'Deep AE'))

# DQL Deep
np.random.seed(42); tf.random.set_seed(42)
_, _, qm_deep = train_dql(X_train, y_train, X_test, y_test, 'deep')
q_out = qm_deep.predict(X_test, verbose=0)
dql_preds  = np.argmax(q_out, axis=1)
dql_scores = q_out[:, 1] - q_out[:, 0]
ext_results.append(get_ext_metrics(y_test, dql_preds, dql_scores, 'DQL Deep'))

import pandas as pd
print('\n' + '='*70)
print(pd.DataFrame(ext_results).to_string(index=False))
print('>>> USE THESE IN TABLE V <<<')
```
