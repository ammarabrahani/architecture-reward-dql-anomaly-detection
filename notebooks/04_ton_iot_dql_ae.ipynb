### Description:
This notebook evaluates anomaly detection models (Isolation Forest, Simple/Deep/Wide Autoencoder, and Deep Q-Learning) on the TON_IoT network dataset. This is the fourth dataset used to validate cross-domain generalisation of the DQL-based framework.

**Dataset:** TON_IoT — University of New South Wales, Australia (2021)

**Attack Types:** Backdoor, DDoS, DoS, Injection, Password, Ransomware, Scanning, XSS, MitM

## Step 1: Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import IsolationForest
from sklearn.metrics import f1_score, precision_score, recall_score, classification_report

import tensorflow as tf
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping

print('All libraries loaded!')
print('TensorFlow:', tf.__version__)
```

## Step 2: Load Dataset

```python
df = pd.read_csv('../data/raw/train_test_network.csv')

print('Shape:', df.shape)
print('\nLabel Distribution:')
print(df['label'].value_counts())
print('\nAnomaly %:', round(df['label'].sum()/len(df)*100, 2), '%')
print('\nAttack Types:')
print(df['type'].value_counts())
df.head()
```

## Step 3: Data Preprocessing

```python
# Drop identifier/string columns that cant be used directly
drop_cols = [
    'src_ip', 'dst_ip',           # IP addresses — identifiers
    'dns_query', 'ssl_subject',    # free text
    'ssl_issuer', 'http_uri',      # free text
    'http_user_agent',             # free text
    'http_orig_mime_types',        # free text
    'http_resp_mime_types',        # free text
    'weird_name', 'weird_addl',    # sparse
    'type'                         # attack category — not needed for binary
]
df_clean = df.drop(columns=drop_cols)

# Label encode remaining categorical columns
cat_cols = df_clean.select_dtypes(include='object').columns.tolist()
cat_cols.remove('label') if 'label' in cat_cols else None
print('Encoding categorical columns:', cat_cols)

le = LabelEncoder()
for col in cat_cols:
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))

# Separate features and labels
X = df_clean.drop(columns=['label']).values.astype(np.float32)
y = df_clean['label'].values

print('\nFeature shape:', X.shape)
print('Missing values:', np.isnan(X).sum())
```

```python
# Min-Max Scaling (same as thesis)
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# 80/20 Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

# Normal data only for Autoencoder
X_train_normal = X_train[y_train == 0]

print('Train set:', X_train.shape, '| Normal:', (y_train==0).sum(), '| Anomaly:', (y_train==1).sum())
print('Test set: ', X_test.shape)
print('Normal-only training samples:', X_train_normal.shape[0])
```

## Step 4: Model 1 — Isolation Forest (Classical Baseline)

```python
print('Training Isolation Forest...')

iso_forest = IsolationForest(n_estimators=100, contamination=0.15, random_state=42)
iso_forest.fit(X_train)

y_pred_if = iso_forest.predict(X_test)
y_pred_if = np.where(y_pred_if == -1, 1, 0)

f1_if = f1_score(y_test, y_pred_if, average='macro')
pr_if = precision_score(y_test, y_pred_if, average='macro')
re_if = recall_score(y_test, y_pred_if, average='macro')

print('\n=== Isolation Forest Results ===')
print(f'Macro F1-Score : {f1_if:.4f}')
print(f'Macro Precision: {pr_if:.4f}')
print(f'Macro Recall   : {re_if:.4f}')
print(classification_report(y_test, y_pred_if, target_names=['Normal', 'Anomaly']))
```

## Step 5: Autoencoder Architectural Comparison — Simple vs Wide vs Deep

```python
input_dim = X_train_normal.shape[1]

def build_simple_autoencoder(input_dim):
    inp = Input(shape=(input_dim,))
    x   = Dense(32, activation='relu')(inp)
    out = Dense(input_dim, activation='linear')(x)
    m   = Model(inp, out)
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

def build_wide_autoencoder(input_dim):
    inp = Input(shape=(input_dim,))
    x   = Dense(128, activation='relu')(inp)
    x   = Dense(128, activation='relu')(x)
    out = Dense(input_dim, activation='linear')(x)
    m   = Model(inp, out)
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

def build_deep_autoencoder(input_dim):
    inp = Input(shape=(input_dim,))
    x   = Dense(64, activation='relu')(inp)
    x   = BatchNormalization()(x)
    x   = Dropout(0.2)(x)
    x   = Dense(32, activation='relu')(x)
    x   = BatchNormalization()(x)
    x   = Dropout(0.2)(x)
    x   = Dense(16, activation='relu')(x)
    x   = Dense(32, activation='relu')(x)
    x   = BatchNormalization()(x)
    x   = Dropout(0.2)(x)
    x   = Dense(64, activation='relu')(x)
    out = Dense(input_dim, activation='linear')(x)
    m   = Model(inp, out)
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

results_arch = {}

for name, model in [
    ('Simple AE', build_simple_autoencoder(input_dim)),
    ('Wide AE',   build_wide_autoencoder(input_dim)),
    ('Deep AE',   build_deep_autoencoder(input_dim))
]:
    print(f'Training {name}...')
    model.fit(
        X_train_normal, X_train_normal,
        epochs=100, batch_size=256,
        validation_split=0.1,
        callbacks=[EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)],
        verbose=0
    )
    recon  = model.predict(X_test, verbose=0)
    errors = np.mean(np.square(X_test - recon), axis=1)
    thresh = np.percentile(errors, 90)
    y_pred = (errors > thresh).astype(int)

    f1 = f1_score(y_test, y_pred, average='macro')
    pr = precision_score(y_test, y_pred, average='macro')
    re = recall_score(y_test, y_pred, average='macro')
    results_arch[name] = {'F1': round(f1,4), 'Precision': round(pr,4), 'Recall': round(re,4)}
    print(f'  {name} — Macro F1: {f1:.4f} | Precision: {pr:.4f} | Recall: {re:.4f}')

f1_ae = results_arch['Deep AE']['F1']
pr_ae = results_arch['Deep AE']['Precision']
re_ae = results_arch['Deep AE']['Recall']

print('\n=== AE Architecture Comparison (TON_IoT) ===')
print(pd.DataFrame(results_arch).T)
```

```python
# AE Architecture Bar Chart
arch_names = list(results_arch.keys())
f1_vals    = [results_arch[k]['F1'] for k in arch_names]
colors     = ['#95a5a6', '#3498db', '#e74c3c']

plt.figure(figsize=(8, 5))
bars = plt.bar(arch_names, f1_vals, color=colors, width=0.5, edgecolor='black', linewidth=0.7)
for bar, score in zip(bars, f1_vals):
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 0.01,
             f'{score:.4f}', ha='center', va='bottom', fontsize=11, fontweight='bold')
plt.ylim(0, 1.1)
plt.ylabel('Macro F1-Score', fontsize=12)
plt.title('Autoencoder Architectural Comparison — TON_IoT', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('ton_iot_ae_comparison.png', dpi=150)
plt.show()
print('Plot saved!')
```

## Step 6: DQL Architectural Comparison — Simple vs Wide vs Deep

```python
# Balanced sample — 500 normal + 500 anomaly
idx_n = np.random.choice(np.where(y_train == 0)[0], size=500, replace=False)
idx_a = np.random.choice(np.where(y_train == 1)[0], size=500, replace=False)
sample_idx = np.concatenate([idx_n, idx_a])
np.random.shuffle(sample_idx)
X_dql = X_train[sample_idx]
y_dql = y_train[sample_idx]
print(f'DQL sample: {len(X_dql)} | Normal: {(y_dql==0).sum()} | Anomaly: {(y_dql==1).sum()}')

def build_dql_network(input_dim, arch='simple'):
    if arch == 'simple':
        m = Sequential([
            Input(shape=(input_dim,)),
            Dense(32, activation='relu'),
            Dense(2,  activation='linear')
        ])
    elif arch == 'deep':
        m = Sequential([
            Input(shape=(input_dim,)),
            Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
            Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
            Dense(16, activation='relu'),
            Dense(32, activation='relu'),
            Dense(2,  activation='linear')
        ])
    elif arch == 'wide':
        m = Sequential([
            Input(shape=(input_dim,)),
            Dense(128, activation='relu'),
            Dense(128, activation='relu'),
            Dense(2,   activation='linear')
        ])
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

dql_arch_results = {}
n_episodes = 20

for arch_name in ['simple', 'deep', 'wide']:
    print(f'\nTraining DQL ({arch_name})...')
    qm = build_dql_network(X_dql.shape[1], arch=arch_name)

    epsilon = 1.0
    f1_eps  = []

    for ep in range(1, n_episodes + 1):
        q_all   = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()

        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            true_label = y_dql[i]
            # Asymmetric dynamic reward
            if action == true_label:
                reward = +1
            elif action == 0 and true_label == 1:
                reward = -2   # missed anomaly — worst case
            else:
                reward = -1   # false alarm
            targets[i][action] = reward

        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)

        preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
        f1    = f1_score(y_test, preds, average='macro')
        f1_eps.append(f1)
        print(f'  Episode {ep:02d}/{n_episodes} — Macro F1: {f1:.4f} | Epsilon: {epsilon:.3f}')

    dql_arch_results[f'DQL {arch_name.capitalize()}'] = {
        'F1': round(f1_eps[-1], 4),
        'Convergence': f1_eps
    }
    print(f'  Final F1 ({arch_name}): {f1_eps[-1]:.4f}')

f1_dql  = dql_arch_results['DQL Deep']['F1']

print('\n=== DQL Architecture Comparison (TON_IoT) ===')
for k, v in dql_arch_results.items():
    print(f'{k}: {v["F1"]}')
```

```python
# DQL Convergence Plot
plt.figure(figsize=(10, 6))
colors_dql = {'DQL Simple': '#95a5a6', 'DQL Deep': '#e74c3c', 'DQL Wide': '#3498db'}

for arch_name, data in dql_arch_results.items():
    plt.plot(range(1, n_episodes+1), data['Convergence'],
             marker='o', label=f"{arch_name} (F1={data['F1']})",
             color=colors_dql[arch_name], linewidth=2)

plt.xlabel('Episode', fontsize=12)
plt.ylabel('Macro F1-Score', fontsize=12)
plt.title('DQL Architecture Comparison — Convergence on TON_IoT', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('ton_iot_dql_convergence.png', dpi=150)
plt.show()
print('Plot saved!')
```

```python
from sklearn.metrics import (confusion_matrix, roc_auc_score, 
                              average_precision_score, f1_score,
                              precision_score, recall_score)
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

def get_metrics(y_true, y_pred, y_score=None, name=''):
    cm = confusion_matrix(y_true, y_pred)
    tn, fp, fn, tp = cm.ravel()
    f1  = f1_score(y_true, y_pred, average='macro')
    pr  = precision_score(y_true, y_pred, average='macro')
    re  = recall_score(y_true, y_pred, average='macro')
    fpr = fp / (fp + tn)
    roc = roc_auc_score(y_true, y_score) if y_score is not None else None
    prc = average_precision_score(y_true, y_score) if y_score is not None else None
    print(f'{name}: F1={f1:.4f} | Prec={pr:.4f} | Rec={re:.4f} | FPR={fpr:.4f} | ROC={roc:.4f} | PR={prc:.4f}')

print('='*65)
print('TON_IoT — Extended Metrics')
print('='*65)

# 1. Isolation Forest
if_scores = -iso_forest.score_samples(X_test)
get_metrics(y_test, y_pred_if, if_scores, 'Isolation Forest')

# 2. Deep AE
model_deep = build_deep_autoencoder(X_train_normal.shape[1])
model_deep.fit(X_train_normal, X_train_normal, epochs=100, batch_size=256,
               validation_split=0.1,
               callbacks=[EarlyStopping(patience=5, restore_best_weights=True)],
               verbose=0)
recon = model_deep.predict(X_test, verbose=0)
ae_scores = np.mean(np.square(X_test - recon), axis=1)
thresh = np.percentile(ae_scores, 90)
y_pred_ae = (ae_scores > thresh).astype(int)
get_metrics(y_test, y_pred_ae, ae_scores, 'Deep AE')

# 3. DQL Deep seed 42
np.random.seed(42)
tf.random.set_seed(42)
idx_n = np.random.choice(np.where(y_train == 0)[0], size=500, replace=False)
idx_a = np.random.choice(np.where(y_train == 1)[0], size=500, replace=False)
X_dql = X_train[np.concatenate([idx_n, idx_a])]
y_dql = y_train[np.concatenate([idx_n, idx_a])]

qm = build_dql_network(X_dql.shape[1], 'deep')
epsilon = 1.0
for ep in range(20):
    q_all = qm.predict(X_dql, verbose=0)
    targets = q_all.copy()
    for i in range(len(X_dql)):
        action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
        reward = +1 if action == y_dql[i] else (-2 if (action==0 and y_dql[i]==1) else -1)
        targets[i][action] = reward
    qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
    epsilon = max(0.05, epsilon * 0.85)

q_out = qm.predict(X_test, verbose=0)
dql_preds = np.argmax(q_out, axis=1)
dql_scores = q_out[:, 1] - q_out[:, 0]
get_metrics(y_test, dql_preds, dql_scores, 'DQL Deep')
```
