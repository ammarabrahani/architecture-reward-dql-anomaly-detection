## Step 1: Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import IsolationForest
from sklearn.metrics import f1_score, precision_score, recall_score, classification_report

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

print('All libraries loaded!')
print('TensorFlow:', tf.__version__)
```

## Step 2: Load Dataset

> **Note:** Place `creditcard.csv` in the same folder as this notebook before running.

```python
df = pd.read_csv('../data/raw/creditcard.csv')

print('Shape:', df.shape)
print('\nLabel Distribution:')
print(df['Class'].value_counts())
print('\nAnomaly %:', round(df['Class'].sum()/len(df)*100, 4), '%')
df.head()
```

## Step 3: Preprocessing

```python
# Drop Time column (same as thesis)
df_clean = df.drop(columns=['Time'])

# Separate features and labels
X = df_clean.drop(columns=['Class']).values.astype(np.float32)
y = df_clean['Class'].values

# Min-Max Scaling
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# 80/20 Stratified Split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print('Train:', X_train.shape, '| Normal:', (y_train==0).sum(), '| Fraud:', (y_train==1).sum())
print('Test: ', X_test.shape,  '| Normal:', (y_test==0).sum(),  '| Fraud:', (y_test==1).sum())
```

## Step 4: Isolation Forest

```python
print('Training Isolation Forest...')

iso = IsolationForest(n_estimators=100, contamination=0.0017, random_state=42)
iso.fit(X_train)

y_pred_if = iso.predict(X_test)
y_pred_if = np.where(y_pred_if == -1, 1, 0)

f1_if = f1_score(y_test, y_pred_if, average='macro')
pr_if = precision_score(y_test, y_pred_if, average='macro')
re_if = recall_score(y_test, y_pred_if, average='macro')

print('\n=== Isolation Forest Results ===')
print(f'Macro F1-Score : {f1_if:.4f}')
print(f'Macro Precision: {pr_if:.4f}')
print(f'Macro Recall   : {re_if:.4f}')
print(classification_report(y_test, y_pred_if, target_names=['Normal', 'Fraud']))
```

## Step 5: DQL Architecture Comparison — Simple vs Wide vs Deep

```python
# Balanced sample — 500 normal + 500 fraud
# Note: only 492 fraud in training set, use all of them
idx_normal = np.where(y_train == 0)[0]
idx_fraud  = np.where(y_train == 1)[0]

n_fraud = len(idx_fraud)
idx_n   = np.random.choice(idx_normal, size=n_fraud, replace=False)
sample_idx = np.concatenate([idx_n, idx_fraud])
np.random.shuffle(sample_idx)

X_dql = X_train[sample_idx]
y_dql = y_train[sample_idx]

print(f'DQL sample: {len(X_dql)} | Normal: {(y_dql==0).sum()} | Fraud: {(y_dql==1).sum()}')

def build_dql_network(input_dim, arch='simple'):
    if arch == 'simple':
        m = Sequential([
            Input(shape=(input_dim,)),
            Dense(32, activation='relu'),
            Dense(2, activation='linear')
        ])
    elif arch == 'deep':
        m = Sequential([
            Input(shape=(input_dim,)),
            Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
            Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
            Dense(16, activation='relu'),
            Dense(32, activation='relu'),
            Dense(2, activation='linear')
        ])
    elif arch == 'wide':
        m = Sequential([
            Input(shape=(input_dim,)),
            Dense(128, activation='relu'),
            Dense(128, activation='relu'),
            Dense(2, activation='linear')
        ])
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

dql_results = {}
n_episodes  = 20

for arch_name in ['simple', 'deep', 'wide']:
    print(f'\nTraining DQL ({arch_name})...')
    qm = build_dql_network(X_dql.shape[1], arch=arch_name)

    epsilon = 1.0
    f1_eps  = []

    for ep in range(1, n_episodes + 1):
        q_all   = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()

        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            true_label = y_dql[i]
            # Asymmetric reward — missing fraud is worse
            if action == true_label:
                reward = +1
            elif action == 0 and true_label == 1:
                reward = -2
            else:
                reward = -1
            targets[i][action] = reward

        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)

        preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
        f1    = f1_score(y_test, preds, average='macro')
        f1_eps.append(f1)
        print(f'  Episode {ep:02d}/{n_episodes} — Macro F1: {f1:.4f} | Epsilon: {epsilon:.3f}')

    dql_results[f'DQL {arch_name.capitalize()}'] = {
        'F1': round(f1_eps[-1], 4),
        'Convergence': f1_eps
    }
    print(f'  Final F1 ({arch_name}): {f1_eps[-1]:.4f}')

print('\n=== DQL Architecture Comparison (Kaggle Credit Card) ===')
for k, v in dql_results.items():
    print(f'{k}: {v["F1"]}')
```

## Step 6: DQL Convergence Plot

```python
plt.figure(figsize=(10, 6))
colors = {'DQL Simple': '#95a5a6', 'DQL Deep': '#e74c3c', 'DQL Wide': '#3498db'}

for name, data in dql_results.items():
    plt.plot(range(1, n_episodes+1), data['Convergence'],
             marker='o', label=f"{name} (F1={data['F1']})",
             color=colors[name], linewidth=2)

plt.axhline(y=f1_if, color='green', linestyle='--',
            linewidth=1.5, label=f'Isolation Forest ({f1_if:.4f})')
plt.axhline(y=0.490, color='orange', linestyle='--',
            linewidth=1.5, label='Deep AE (0.490)')

plt.xlabel('Episode', fontsize=12)
plt.ylabel('Macro F1-Score', fontsize=12)
plt.title('DQL Convergence — Kaggle Credit Card Fraud Dataset', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('kaggle_dql_convergence.png', dpi=150)
plt.show()
print('Plot saved!')
```

## Step 8: Symmetric vs Asymmetric Reward Comparison

**Giovani's suggestion:** Compare symmetric reward (+1/-1) vs asymmetric reward (+1/-2/-1)
to quantify the benefit of penalising missed anomalies more heavily.

```python
print('Comparing Symmetric vs Asymmetric Reward on Kaggle Credit Card...')
print('Using DQL Deep architecture (best performer)\n')

# Balanced sample — all fraud samples
idx_fraud  = np.where(y_train == 1)[0]
idx_normal = np.where(y_train == 0)[0]
n_fraud    = len(idx_fraud)
idx_n      = np.random.choice(idx_normal, size=n_fraud, replace=False)
sample_idx = np.concatenate([idx_n, idx_fraud])
np.random.shuffle(sample_idx)
X_dql = X_train[sample_idx]
y_dql = y_train[sample_idx]
print(f'Sample: {len(X_dql)} | Normal: {(y_dql==0).sum()} | Fraud: {(y_dql==1).sum()}')

n_episodes = 20
reward_schemes = {
    'Asymmetric (+1/-2/-1)': {'miss_anomaly': -2, 'false_alarm': -1},
    'Symmetric  (+1/-1/-1)': {'miss_anomaly': -1, 'false_alarm': -1},
}

reward_results = {}

for scheme_name, rewards in reward_schemes.items():
    print(f'\nTraining: {scheme_name}')

    qm = Sequential([
        Input(shape=(X_dql.shape[1],)),
        Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(16, activation='relu'),
        Dense(32, activation='relu'),
        Dense(2,  activation='linear')
    ])
    qm.compile(optimizer=Adam(0.001), loss='mse')

    epsilon = 1.0
    f1_eps  = []

    for ep in range(1, n_episodes + 1):
        q_all   = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()

        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            true_label = y_dql[i]

            if action == true_label:
                reward = +1
            elif action == 0 and true_label == 1:
                reward = rewards['miss_anomaly']
            else:
                reward = rewards['false_alarm']

            targets[i][action] = reward

        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)

        preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
        f1    = f1_score(y_test, preds, average='macro')
        f1_eps.append(f1)
        print(f'  Episode {ep:02d}/{n_episodes} — Macro F1: {f1:.4f}')

    reward_results[scheme_name] = {
        'F1': round(f1_eps[-1], 4),
        'Convergence': f1_eps
    }
    print(f'  Final F1: {f1_eps[-1]:.4f}')

asym_f1 = reward_results['Asymmetric (+1/-2/-1)']['F1']
sym_f1  = reward_results['Symmetric  (+1/-1/-1)']['F1']
improvement = round(((asym_f1 - sym_f1) / sym_f1) * 100, 2)

print('\n' + '='*55)
print('Kaggle Credit Card — Reward Scheme Comparison')
print('='*55)
print(f'Symmetric  (+1/-1/-1) : {sym_f1}')
print(f'Asymmetric (+1/-2/-1) : {asym_f1}')
print(f'Improvement           : {improvement}%')
print('='*55)
```

```python
import numpy as np
import tensorflow as tf
from sklearn.metrics import f1_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

seeds = [42, 0, 7, 123, 2024]
n_episodes = 20

def run_reward_kaggle(X_train, y_train, X_test, y_test, seed, asymmetric=True):
    np.random.seed(seed)
    tf.random.set_seed(seed)
    
    # Use all fraud + equal normal
    idx_fraud  = np.where(y_train == 1)[0]
    idx_normal = np.where(y_train == 0)[0]
    idx_n = np.random.choice(idx_normal, size=len(idx_fraud), replace=False)
    sample_idx = np.concatenate([idx_n, idx_fraud])
    np.random.shuffle(sample_idx)
    X_dql = X_train[sample_idx]
    y_dql = y_train[sample_idx]

    qm = Sequential([
        Input(shape=(X_dql.shape[1],)),
        Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(16, activation='relu'), Dense(32, activation='relu'),
        Dense(2, activation='linear')
    ])
    qm.compile(optimizer=Adam(0.001), loss='mse')
    epsilon = 1.0
    for ep in range(n_episodes):
        q_all = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()
        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            if action == y_dql[i]: reward = +1
            elif action == 0 and y_dql[i] == 1: reward = -2 if asymmetric else -1
            else: reward = -1
            targets[i][action] = reward
        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)
    preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
    return f1_score(y_test, preds, average='macro')

asym_f1s = []
sym_f1s  = []
for seed in seeds:
    a = run_reward_kaggle(X_train, y_train, X_test, y_test, seed, asymmetric=True)
    s = run_reward_kaggle(X_train, y_train, X_test, y_test, seed, asymmetric=False)
    asym_f1s.append(a)
    sym_f1s.append(s)
    print(f'Seed {seed} → Asymmetric: {a:.4f} | Symmetric: {s:.4f}')

asym_mean = np.mean(asym_f1s)
sym_mean  = np.mean(sym_f1s)
improvement = ((asym_mean - sym_mean) / sym_mean) * 100

print(f'\nSymmetric  : {sym_mean:.4f} +/- {np.std(sym_f1s):.4f}')
print(f'Asymmetric : {asym_mean:.4f} +/- {np.std(asym_f1s):.4f}')
print(f'Improvement: {improvement:.2f}%')
print('\n>>> USE THESE IN TABLE IV <<<')
```
