### Description:
This notebook evaluates the performance of a basic feedforward Autoencoder for anomaly detection on a synthetic dataset generated using `make_blobs`. Anomalies are introduced manually via uniformly distributed noise. The Autoencoder is trained only on normal data and attempts to reconstruct inputs. Anomalies are identified based on high reconstruction error (MSE).

```python
# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_moons
from sklearn.metrics import f1_score
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.optimizers import Adam
```

## Step 1: Define Autoencoder Model

```python

# Define helper function to build Autoencoder
def build_autoencoder(input_dim):
    input_layer = Input(shape=(input_dim,))
    encoder = Dense(8, activation="relu")(input_layer)
    decoder = Dense(input_dim, activation="linear")(encoder)
    autoencoder = Model(inputs=input_layer, outputs=decoder)
    autoencoder.compile(optimizer=Adam(learning_rate=0.001), loss='mse')
    return autoencoder
```

## Step 2: Generate Dataset

```python

# Generate datasets
n_samples = 300
outliers_fraction = 0.15
n_outliers = int(outliers_fraction * n_samples)
n_inliers = n_samples
blobs_params = dict(random_state=0, n_samples=n_inliers, n_features=2)
datasets = [
    make_blobs(centers=[[0, 0], [0, 0]], cluster_std=0.5, **blobs_params)[0],
]
```

## Step 3: Train Autoencoder and Evaluate

```python

# Evaluate Autoencoder on each dataset
results = []

for i, X in enumerate(datasets):
    rng = np.random.RandomState(42)
    X_outliers = rng.uniform(low=-6, high=6, size=(n_outliers, 2))
    X_full = np.concatenate([X, X_outliers], axis=0)
    y_true = np.concatenate([np.zeros(n_inliers), np.ones(n_outliers)])

    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_full)

    model = build_autoencoder(input_dim=2)
    model.fit(X_scaled[y_true == 0], X_scaled[y_true == 0],
              epochs=20, batch_size=16, verbose=0)

    reconstructions = model.predict(X_scaled)
    mse = np.mean(np.square(X_scaled - reconstructions), axis=1)
    threshold = np.percentile(mse, 100 * (1 - outliers_fraction))
    y_pred = (mse > threshold).astype(int)

    f1 = f1_score(y_true, y_pred, average='macro')
    acc = np.mean(y_true == y_pred)

    results.append({
        'Model': 'Autoencoder',
        'Macro F1-Score': round(f1, 4),
        'Accuracy': round(acc, 4)
    })
```

## Step 4: Print Results

```python

df = pd.DataFrame(results)
print(df.to_string(index=False))
```

## Step 5: Symmetric vs Asymmetric Reward Comparison on Synthetic Data

**Purpose:** Complete Table IV by adding Synthetic dataset results.

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_classification
from sklearn.metrics import f1_score
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

# Generate synthetic dataset (same as thesis)
X_syn, y_syn = make_classification(
    n_samples=5000, n_features=20,
    n_informative=15, n_redundant=5,
    weights=[0.85, 0.15],  # 15% anomaly
    random_state=42
)

# Scale and split
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X_syn)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_syn, test_size=0.2, random_state=42, stratify=y_syn
)

# Balanced sample
idx_n = np.where(y_train == 0)[0]
idx_a = np.where(y_train == 1)[0]
idx_n_s = np.random.choice(idx_n, size=len(idx_a), replace=False)
sample_idx = np.concatenate([idx_n_s, idx_a])
np.random.shuffle(sample_idx)
X_dql = X_train[sample_idx]
y_dql = y_train[sample_idx]
print(f'Sample: {len(X_dql)} | Normal: {(y_dql==0).sum()} | Anomaly: {(y_dql==1).sum()}')

n_episodes = 20
reward_schemes = {
    'Asymmetric (+1/-2/-1)': {'miss_anomaly': -2, 'false_alarm': -1},
    'Symmetric  (+1/-1/-1)': {'miss_anomaly': -1, 'false_alarm': -1},
}

reward_results = {}

for scheme_name, rewards in reward_schemes.items():
    print(f'\nTraining: {scheme_name}')

    qm = Sequential([
        Input(shape=(X_dql.shape[1],)),
        Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(16, activation='relu'),
        Dense(32, activation='relu'),
        Dense(2,  activation='linear')
    ])
    qm.compile(optimizer=Adam(0.001), loss='mse')

    epsilon = 1.0
    f1_eps  = []

    for ep in range(1, n_episodes + 1):
        q_all   = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()

        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            true_label = y_dql[i]
            if action == true_label:
                reward = +1
            elif action == 0 and true_label == 1:
                reward = rewards['miss_anomaly']
            else:
                reward = rewards['false_alarm']
            targets[i][action] = reward

        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)

        preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
        f1    = f1_score(y_test, preds, average='macro')
        f1_eps.append(f1)
        print(f'  Episode {ep:02d}/{n_episodes} — Macro F1: {f1:.4f}')

    reward_results[scheme_name] = {
        'F1': round(f1_eps[-1], 4),
        'Convergence': f1_eps
    }
    print(f'  Final F1: {f1_eps[-1]:.4f}')

asym_f1 = reward_results['Asymmetric (+1/-2/-1)']['F1']
sym_f1  = reward_results['Symmetric  (+1/-1/-1)']['F1']
improvement = round(((asym_f1 - sym_f1) / sym_f1) * 100, 2)

print('\n' + '='*55)
print('Synthetic — Reward Scheme Comparison')
print('='*55)
print(f'Symmetric  (+1/-1/-1) : {sym_f1}')
print(f'Asymmetric (+1/-2/-1) : {asym_f1}')
print(f'Improvement           : {improvement}%')
print('='*55)
print('\n>>> ADD THESE TO TABLE IV IN main.tex <<<')
```

## Step 6: Multiple Random Seeds — Statistical Validation

**Reviewer Comment 2:** Run DQL Deep with 5 different random seeds.

Seeds: 42, 0, 7, 123, 2024

```python
import tensorflow as tf
from sklearn.datasets import make_classification
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, precision_score, recall_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

seeds = [42, 0, 7, 123, 2024]
n_episodes = 20
f1_scores  = []
pr_scores  = []
re_scores  = []

print('Running DQL Deep on Synthetic data with 5 seeds...\n')

for seed in seeds:
    np.random.seed(seed)
    tf.random.set_seed(seed)

    # Generate synthetic dataset with this seed
    X_syn, y_syn = make_classification(
        n_samples=5000, n_features=20,
        n_informative=15, n_redundant=5,
        weights=[0.85, 0.15],
        random_state=seed
    )
    scaler = MinMaxScaler()
    X_scaled = scaler.fit_transform(X_syn)
    X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
        X_scaled, y_syn, test_size=0.2, random_state=seed, stratify=y_syn
    )

    # Balanced sample
    idx_n = np.where(y_train_s == 0)[0]
    idx_a = np.where(y_train_s == 1)[0]
    idx_n_s = np.random.choice(idx_n, size=len(idx_a), replace=False)
    sample_idx = np.concatenate([idx_n_s, idx_a])
    np.random.shuffle(sample_idx)
    X_dql = X_train_s[sample_idx]
    y_dql = y_train_s[sample_idx]

    # Build DQL Deep
    qm = Sequential([
        Input(shape=(X_dql.shape[1],)),
        Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(16, activation='relu'),
        Dense(32, activation='relu'),
        Dense(2,  activation='linear')
    ])
    qm.compile(optimizer=Adam(0.001), loss='mse')

    epsilon = 1.0

    for ep in range(1, n_episodes + 1):
        q_all   = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()
        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            true_label = y_dql[i]
            if action == true_label:
                reward = +1
            elif action == 0 and true_label == 1:
                reward = -2
            else:
                reward = -1
            targets[i][action] = reward
        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)

    preds = np.argmax(qm.predict(X_test_s, verbose=0), axis=1)
    f1 = f1_score(y_test_s, preds, average='macro')
    pr = precision_score(y_test_s, preds, average='macro')
    re = recall_score(y_test_s, preds, average='macro')
    f1_scores.append(f1)
    pr_scores.append(pr)
    re_scores.append(re)
    print(f'Seed {seed:4d} → F1: {f1:.4f} | Precision: {pr:.4f} | Recall: {re:.4f}')

print('\n' + '='*55)
print('Synthetic — DQL Deep (5 Seeds Summary)')
print('='*55)
print(f'Macro F1   : {np.mean(f1_scores):.4f} ± {np.std(f1_scores):.4f}')
print(f'Precision  : {np.mean(pr_scores):.4f} ± {np.std(pr_scores):.4f}')
print(f'Recall     : {np.mean(re_scores):.4f} ± {np.std(re_scores):.4f}')
print('='*55)
print('\n>>> USE THESE VALUES IN TABLE V OF main.tex <<<')
```

## Step 7: DQL Convergence Plot — Synthetic Data

Generating convergence plot for camera-ready paper.

```python
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf
from sklearn.datasets import make_classification
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

# Generate synthetic data
X_syn, y_syn = make_classification(
    n_samples=5000, n_features=20,
    n_informative=15, n_redundant=5,
    weights=[0.85, 0.15], random_state=42
)
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X_syn)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y_syn, test_size=0.2, random_state=42, stratify=y_syn
)

# Balanced sample
idx_n = np.where(y_train == 0)[0]
idx_a = np.where(y_train == 1)[0]
idx_n_s = np.random.choice(idx_n, size=len(idx_a), replace=False)
sample_idx = np.concatenate([idx_n_s, idx_a])
np.random.shuffle(sample_idx)
X_dql = X_train[sample_idx]
y_dql = y_train[sample_idx]

def build_dql_network(input_dim, arch='simple'):
    if arch == 'simple':
        m = Sequential([Input(shape=(input_dim,)), Dense(32, activation='relu'), Dense(2, activation='linear')])
    elif arch == 'deep':
        m = Sequential([Input(shape=(input_dim,)), Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2), Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2), Dense(16, activation='relu'), Dense(32, activation='relu'), Dense(2, activation='linear')])
    elif arch == 'wide':
        m = Sequential([Input(shape=(input_dim,)), Dense(128, activation='relu'), Dense(128, activation='relu'), Dense(2, activation='linear')])
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

dql_arch_results = {}
n_episodes = 20

for arch_name in ['simple', 'deep', 'wide']:
    print(f'Training DQL ({arch_name})...')
    qm = build_dql_network(X_dql.shape[1], arch=arch_name)
    epsilon = 1.0
    f1_eps = []
    for ep in range(1, n_episodes + 1):
        q_all = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()
        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            reward = +1 if action == y_dql[i] else (-2 if (action==0 and y_dql[i]==1) else -1)
            targets[i][action] = reward
        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)
        preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
        f1 = f1_score(y_test, preds, average='macro')
        f1_eps.append(f1)
        print(f'  Episode {ep:02d}/{n_episodes} — F1: {f1:.4f}')
    dql_arch_results[f'DQL {arch_name.capitalize()}'] = {'F1': round(f1_eps[-1], 4), 'Convergence': f1_eps}
    print(f'Final F1: {f1_eps[-1]:.4f}\n')

# Plot
plt.figure(figsize=(10, 6))
colors_dql = {'DQL Simple': '#95a5a6', 'DQL Deep': '#e74c3c', 'DQL Wide': '#3498db'}
for arch_name, data in dql_arch_results.items():
    plt.plot(range(1, n_episodes+1), data['Convergence'], marker='o',
             label=f"{arch_name} (F1={data['F1']})", color=colors_dql[arch_name], linewidth=2)
plt.axhline(y=0.934, color='green', linestyle='--', linewidth=1.5, label='Isolation Forest (0.934)')
plt.axhline(y=0.922, color='orange', linestyle='--', linewidth=1.5, label='Deep AE (0.922)')
plt.xlabel('Episode', fontsize=12)
plt.ylabel('Macro F1-Score', fontsize=12)
plt.title('DQL Architecture Comparison — Convergence on Synthetic Data', fontsize=12)
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('synthetic_dql_convergence.png', dpi=150)
plt.show()
print('Plot saved: synthetic_dql_convergence.png')
```

## Reward Comparison: Symmetric vs Asymmetric (5 Seeds) — Synthetic

```python
import numpy as np
import tensorflow as tf
from sklearn.metrics import f1_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

seeds = [42, 0, 7, 123, 2024]
n_episodes = 20

def run_reward(X_train, y_train, X_test, y_test, seed, asymmetric=True):
    np.random.seed(seed)
    tf.random.set_seed(seed)
    idx_n = np.random.choice(np.where(y_train == 0)[0], size=500, replace=False)
    idx_a = np.random.choice(np.where(y_train == 1)[0], size=500, replace=False)
    sample_idx = np.concatenate([idx_n, idx_a])
    np.random.shuffle(sample_idx)
    X_dql = X_train[sample_idx]
    y_dql = y_train[sample_idx]
    qm = Sequential([
        Input(shape=(X_dql.shape[1],)),
        Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(16, activation='relu'), Dense(32, activation='relu'),
        Dense(2, activation='linear')
    ])
    qm.compile(optimizer=Adam(0.001), loss='mse')
    epsilon = 1.0
    for ep in range(n_episodes):
        q_all = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()
        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            if action == y_dql[i]: reward = +1
            elif action == 0 and y_dql[i] == 1: reward = -2 if asymmetric else -1
            else: reward = -1
            targets[i][action] = reward
        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)
    preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
    return f1_score(y_test, preds, average='macro')

asym_f1s = []
sym_f1s  = []
for seed in seeds:
    a = run_reward(X_train, y_train, X_test, y_test, seed, asymmetric=True)
    s = run_reward(X_train, y_train, X_test, y_test, seed, asymmetric=False)
    asym_f1s.append(a)
    sym_f1s.append(s)
    print(f'Seed {seed} → Asymmetric: {a:.4f} | Symmetric: {s:.4f}')

asym_mean = np.mean(asym_f1s)
sym_mean  = np.mean(sym_f1s)
improvement = ((asym_mean - sym_mean) / sym_mean) * 100

print(f'\nSymmetric  : {sym_mean:.4f} +/- {np.std(sym_f1s):.4f}')
print(f'Asymmetric : {asym_mean:.4f} +/- {np.std(asym_f1s):.4f}')
print(f'Improvement: {improvement:.2f}%')
print('\n>>> USE THESE IN TABLE IV <<<')
```
