## Step 1: Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import IsolationForest
from sklearn.metrics import f1_score, precision_score, recall_score, classification_report, confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Model, Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping


print('All libraries loaded successfully!')
print('TensorFlow version:', tf.__version__)
```

## Step 2: Load Dataset

```python
# Load training and testing sets
df_train = pd.read_parquet('../data/raw/UNSW_NB15_training-set.parquet')
df_test  = pd.read_parquet('../data/raw/UNSW_NB15_testing-set.parquet')

# Combine for full dataset (we will do our own split)
df = pd.concat([df_train, df_test], ignore_index=True)

print('Combined Dataset Shape:', df.shape)
print('\nLabel Distribution:')
print(df['label'].value_counts())
print('\nAnomaly %:', round(df['label'].sum() / len(df) * 100, 2), '%')
print('\nFirst 5 rows:')
df.head()
```

## Step 3: Data Preprocessing

```python
# Drop non-numeric / identifier columns
# 'attack_cat' = attack category (too specific, not needed for binary classification)
# 'proto', 'service', 'state' = categorical — drop for simplicity (same approach as thesis)
drop_cols = ['proto', 'service', 'state', 'attack_cat']
df_clean = df.drop(columns=drop_cols)

print('Columns after dropping categoricals:', df_clean.shape[1])
print('Missing values:', df_clean.isnull().sum().sum())

# Separate features and labels
X = df_clean.drop(columns=['label']).values.astype(np.float32)
y = df_clean['label'].values

print('\nFeature shape:', X.shape)
print('Label shape:', y.shape)
```

```python
# Apply Min-Max Scaling (same as thesis approach)
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# 80/20 Train-Test Split with stratification (same as thesis)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print('Train set:', X_train.shape, '| Label dist:', np.bincount(y_train))
print('Test set: ', X_test.shape,  '| Label dist:', np.bincount(y_test))

# Normal data only (for Autoencoder training)
X_train_normal = X_train[y_train == 0]
print('\nNormal-only training samples:', X_train_normal.shape[0])
```

## Step 4: Model 1 — Isolation Forest (Classical Baseline)

```python
print('Training Isolation Forest...')

iso_forest = IsolationForest(
    n_estimators=100,
    contamination=0.15,
    random_state=42
)
iso_forest.fit(X_train)

# Isolation Forest returns -1 for anomaly, 1 for normal
# Convert to 0 = normal, 1 = anomaly
y_pred_if = iso_forest.predict(X_test)
y_pred_if = np.where(y_pred_if == -1, 1, 0)

f1_if = f1_score(y_test, y_pred_if, average='macro')
pr_if = precision_score(y_test, y_pred_if, average='macro')
re_if = recall_score(y_test, y_pred_if, average='macro')

print('\n=== Isolation Forest Results ===')
print(f'Macro F1-Score : {f1_if:.4f}')
print(f'Macro Precision: {pr_if:.4f}')
print(f'Macro Recall   : {re_if:.4f}')
print('\nClassification Report:')
print(classification_report(y_test, y_pred_if, target_names=['Normal', 'Anomaly']))
```

## Step 5: Model 2 — Deep Autoencoder (Best Architecture from Thesis)

```python
def build_deep_autoencoder(input_dim):
    """
    Deep Autoencoder — same architecture as thesis (Use Case 2 winner)
    6 layers with Dropout + BatchNormalization
    """
    input_layer = Input(shape=(input_dim,))

    # Encoder
    x = Dense(64, activation='relu')(input_layer)
    x = BatchNormalization()(x)
    x = Dropout(0.2)(x)
    x = Dense(32, activation='relu')(x)
    x = BatchNormalization()(x)
    x = Dropout(0.2)(x)
    x = Dense(16, activation='relu')(x)  # Bottleneck

    # Decoder
    x = Dense(32, activation='relu')(x)
    x = BatchNormalization()(x)
    x = Dropout(0.2)(x)
    x = Dense(64, activation='relu')(x)
    output_layer = Dense(input_dim, activation='linear')(x)

    autoencoder = Model(inputs=input_layer, outputs=output_layer)
    autoencoder.compile(optimizer=Adam(learning_rate=0.001), loss='mse')
    return autoencoder

input_dim = X_train_normal.shape[1]
ae_model = build_deep_autoencoder(input_dim)
ae_model.summary()
```

```python
print('Training Deep Autoencoder on normal data only...')

early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

history = ae_model.fit(
    X_train_normal, X_train_normal,
    epochs=100,
    batch_size=256,
    validation_split=0.1,
    callbacks=[early_stop],
    verbose=1
)

print('Training complete!')
```

```python
# Evaluate Autoencoder
reconstructions = ae_model.predict(X_test, verbose=0)
mse_errors = np.mean(np.square(X_test - reconstructions), axis=1)

# Use 90th percentile threshold (same as thesis)
threshold = np.percentile(mse_errors, 90)
print(f'Reconstruction threshold (90th percentile): {threshold:.6f}')

y_pred_ae = (mse_errors > threshold).astype(int)

f1_ae = f1_score(y_test, y_pred_ae, average='macro')
pr_ae = precision_score(y_test, y_pred_ae, average='macro')
re_ae = recall_score(y_test, y_pred_ae, average='macro')

print('\n=== Deep Autoencoder Results ===')
print(f'Macro F1-Score : {f1_ae:.4f}')
print(f'Macro Precision: {pr_ae:.4f}')
print(f'Macro Recall   : {re_ae:.4f}')
print('\nClassification Report:')
print(classification_report(y_test, y_pred_ae, target_names=['Normal', 'Anomaly']))
```

## Step 5b: Architectural Comparison — Simple, Deep, Wide Autoencoder

```python
# ── Simple Autoencoder ──────────────────────────────────────────
def build_simple_autoencoder(input_dim):
    inp = Input(shape=(input_dim,))
    x   = Dense(32, activation='relu')(inp)
    out = Dense(input_dim, activation='linear')(x)
    m   = Model(inp, out)
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

# ── Wide Autoencoder ────────────────────────────────────────────
def build_wide_autoencoder(input_dim):
    inp = Input(shape=(input_dim,))
    x   = Dense(128, activation='relu')(inp)
    x   = Dense(128, activation='relu')(x)
    out = Dense(input_dim, activation='linear')(x)
    m   = Model(inp, out)
    m.compile(optimizer=Adam(0.001), loss='mse')
    return m

input_dim = X_train_normal.shape[1]
results_arch = {}

for name, model in [
    ('Simple AE', build_simple_autoencoder(input_dim)),
    ('Wide AE',   build_wide_autoencoder(input_dim)),
    ('Deep AE',   build_deep_autoencoder(input_dim))
]:
    print(f'Training {name}...')
    model.fit(
        X_train_normal, X_train_normal,
        epochs=100, batch_size=256,
        validation_split=0.1,
        callbacks=[EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)],
        verbose=0
    )
    recon  = model.predict(X_test, verbose=0)
    errors = np.mean(np.square(X_test - recon), axis=1)
    thresh = np.percentile(errors, 90)
    y_pred = (errors > thresh).astype(int)

    f1 = f1_score(y_test, y_pred, average='macro')
    pr = precision_score(y_test, y_pred, average='macro')
    re = recall_score(y_test, y_pred, average='macro')
    results_arch[name] = {'F1': round(f1,4), 'Precision': round(pr,4), 'Recall': round(re,4)}
    print(f'  {name} — Macro F1: {f1:.4f} | Precision: {pr:.4f} | Recall: {re:.4f}')

print('\n=== Autoencoder Architecture Comparison (UNSW-NB15) ===')
df_arch = pd.DataFrame(results_arch).T
print(df_arch)
```

## Step 5d: DQL Architecture Comparison — Simple, Deep, Wide

```python
# Balanced sample
idx_n = np.random.choice(np.where(y_train == 0)[0], size=500, replace=False)
idx_a = np.random.choice(np.where(y_train == 1)[0], size=500, replace=False)
sample_idx = np.concatenate([idx_n, idx_a])
np.random.shuffle(sample_idx)
X_dql = X_train[sample_idx]
y_dql = y_train[sample_idx]

# Simulate labels
center = np.median(X_dql, axis=0)
dists  = np.linalg.norm(X_dql - center, axis=1)
y_sim  = y_dql  # use real labels

def build_dql_network(input_dim, arch='simple'):
    if arch == 'simple':
        return Sequential([
            Input(shape=(input_dim,)),
            Dense(32, activation='relu'),
            Dense(2,  activation='linear')
        ])
    elif arch == 'deep':
        return Sequential([
            Input(shape=(input_dim,)),
            Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
            Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
            Dense(16, activation='relu'),
            Dense(32, activation='relu'),
            Dense(2,  activation='linear')
        ])
    elif arch == 'wide':
        return Sequential([
            Input(shape=(input_dim,)),
            Dense(128, activation='relu'),
            Dense(128, activation='relu'),
            Dense(2,   activation='linear')
        ])

dql_arch_results = {}
n_episodes = 20

for arch_name in ['simple', 'deep', 'wide']:
    print(f'\nTraining DQL ({arch_name})...')
    qm = build_dql_network(X_dql.shape[1], arch=arch_name)
    qm.compile(optimizer=Adam(0.001), loss='mse')

    epsilon = 1.0
    f1_eps  = []

    for ep in range(1, n_episodes + 1):
        q_all   = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()
        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            reward = +1 if action == y_sim[i] else (-2 if (action==0 and y_sim[i]==1) else -1)
            targets[i][action] = reward
        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)

        preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
        f1    = f1_score(y_test, preds, average='macro')
        f1_eps.append(f1)
        print(f'  Episode {ep:02d}/{n_episodes} — Macro F1: {f1:.4f}')

    dql_arch_results[f'DQL {arch_name.capitalize()}'] = {
        'F1': round(f1_eps[-1], 4),
        'Convergence': f1_eps
    }
    print(f'Final F1 ({arch_name}): {f1_eps[-1]:.4f}')

print('\n=== DQL Architecture Comparison (UNSW-NB15) ===')
for k, v in dql_arch_results.items():
    print(f'{k}: {v["F1"]}')
```

## Step 6: Model 3 — Deep Q-Learning (DQL)

```python
print('Training DQL Agent over 20 episodes...')
print('This may take a few minutes.\n')

# Balanced sample — 500 normal + 500 anomaly
idx_normal  = np.where(y_train == 0)[0]
idx_anomaly = np.where(y_train == 1)[0]
idx_n = np.random.choice(idx_normal,  size=500, replace=False)
idx_a = np.random.choice(idx_anomaly, size=500, replace=False)
sample_idx  = np.concatenate([idx_n, idx_a])
np.random.shuffle(sample_idx)
X_train_dql = X_train[sample_idx]
y_train_dql = y_train[sample_idx]
print(f'DQL sample: {len(X_train_dql)} | Normal: {(y_train_dql==0).sum()} | Anomaly: {(y_train_dql==1).sum()}')

f1_per_episode = []
n_episodes = 20
input_dim = X_train_dql.shape[1]

# Build Q-network
q_model = Sequential([
    Input(shape=(input_dim,)),
    Dense(64, activation='relu'),
    BatchNormalization(),
    Dropout(0.2),
    Dense(32, activation='relu'),
    BatchNormalization(),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(32, activation='relu'),
    Dense(2, activation='linear')
])
q_model.compile(optimizer=Adam(learning_rate=0.001), loss='mse')

epsilon = 1.0
gamma   = 0.95

for episode in range(1, n_episodes + 1):
    # BATCH approach — no row-by-row predict
    q_vals_all = q_model.predict(X_train_dql, verbose=0)
    targets = q_vals_all.copy()

    for i in range(len(X_train_dql)):
        action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_vals_all[i])
        true_label = y_train_dql[i]

        # Asymmetric reward
        if action == true_label:
            reward = +1
        elif action == 0 and true_label == 1:  # missed anomaly
            reward = -2
        else:
            reward = -1

        targets[i][action] = reward

    # Batch train
    q_model.fit(X_train_dql, targets, epochs=5, batch_size=64, verbose=0)
    epsilon = max(0.05, epsilon * 0.85)

    # Evaluate
    preds = np.argmax(q_model.predict(X_test, verbose=0), axis=1)
    f1 = f1_score(y_test, preds, average='macro')
    f1_per_episode.append(f1)
    print(f'Episode {episode:02d}/{n_episodes} — Macro F1: {f1:.4f} | Epsilon: {epsilon:.3f}')

print('\nDQL Training Complete!')
print(f'Final DQL Macro F1-Score (Episode 20): {f1_per_episode[-1]:.4f}')
```

## Step 9: Symmetric vs Asymmetric Reward Comparison

**Giovani's suggestion:** Compare symmetric reward (+1/-1) vs asymmetric reward (+1/-2/-1)
to quantify the benefit of penalising missed anomalies more heavily.

```python
print('Comparing Symmetric vs Asymmetric Reward on UNSW-NB15...')
print('Using DQL Deep architecture (best performer)\n')

# Balanced sample — same as before
idx_n = np.random.choice(np.where(y_train == 0)[0], size=500, replace=False)
idx_a = np.random.choice(np.where(y_train == 1)[0], size=500, replace=False)
sample_idx = np.concatenate([idx_n, idx_a])
np.random.shuffle(sample_idx)
X_dql = X_train[sample_idx]
y_dql = y_train[sample_idx]

n_episodes = 20
reward_schemes = {
    'Asymmetric (+1/-2/-1)': {'miss_anomaly': -2, 'false_alarm': -1},
    'Symmetric  (+1/-1/-1)': {'miss_anomaly': -1, 'false_alarm': -1},
}

reward_results = {}

for scheme_name, rewards in reward_schemes.items():
    print(f'Training: {scheme_name}')

    # Build fresh DQL Deep model
    qm = Sequential([
        Input(shape=(X_dql.shape[1],)),
        Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(16, activation='relu'),
        Dense(32, activation='relu'),
        Dense(2,  activation='linear')
    ])
    qm.compile(optimizer=Adam(0.001), loss='mse')

    epsilon = 1.0
    f1_eps  = []

    for ep in range(1, n_episodes + 1):
        q_all   = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()

        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            true_label = y_dql[i]

            if action == true_label:
                reward = +1
            elif action == 0 and true_label == 1:
                reward = rewards['miss_anomaly']
            else:
                reward = rewards['false_alarm']

            targets[i][action] = reward

        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)

        preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
        f1    = f1_score(y_test, preds, average='macro')
        f1_eps.append(f1)
        print(f'  Episode {ep:02d}/{n_episodes} — Macro F1: {f1:.4f}')

    reward_results[scheme_name] = {
        'F1': round(f1_eps[-1], 4),
        'Convergence': f1_eps
    }
    print(f'  Final F1: {f1_eps[-1]:.4f}\n')

# Summary
asym_f1 = reward_results['Asymmetric (+1/-2/-1)']['F1']
sym_f1  = reward_results['Symmetric  (+1/-1/-1)']['F1']
improvement = round(((asym_f1 - sym_f1) / sym_f1) * 100, 2)

print('='*55)
print('UNSW-NB15 — Reward Scheme Comparison')
print('='*55)
print(f'Symmetric  (+1/-1/-1) : {sym_f1}')
print(f'Asymmetric (+1/-2/-1) : {asym_f1}')
print(f'Improvement           : {improvement}%')
print('='*55)
```

```python
# Convergence comparison plot
plt.figure(figsize=(10, 6))
colors = {
    'Asymmetric (+1/-2/-1)': '#e74c3c',
    'Symmetric  (+1/-1/-1)': '#3498db'
}

for name, data in reward_results.items():
    plt.plot(range(1, n_episodes+1), data['Convergence'],
             marker='o', label=f"{name} (F1={data['F1']})",
             color=colors[name], linewidth=2)

plt.xlabel('Episode', fontsize=12)
plt.ylabel('Macro F1-Score', fontsize=12)
plt.title('Symmetric vs Asymmetric Reward — UNSW-NB15', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('unsw_reward_comparison.png', dpi=150)
plt.show()
print('Plot saved: unsw_reward_comparison.png')
```

## Reward Comparison: Symmetric vs Asymmetric (5 Seeds) — UNSW

```python
import numpy as np
import tensorflow as tf
from sklearn.metrics import f1_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam

seeds = [42, 0, 7, 123, 2024]
n_episodes = 20

def run_reward(X_train, y_train, X_test, y_test, seed, asymmetric=True):
    np.random.seed(seed)
    tf.random.set_seed(seed)
    idx_n = np.random.choice(np.where(y_train == 0)[0], size=500, replace=False)
    idx_a = np.random.choice(np.where(y_train == 1)[0], size=500, replace=False)
    sample_idx = np.concatenate([idx_n, idx_a])
    np.random.shuffle(sample_idx)
    X_dql = X_train[sample_idx]
    y_dql = y_train[sample_idx]
    qm = Sequential([
        Input(shape=(X_dql.shape[1],)),
        Dense(64, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(32, activation='relu'), BatchNormalization(), Dropout(0.2),
        Dense(16, activation='relu'), Dense(32, activation='relu'),
        Dense(2, activation='linear')
    ])
    qm.compile(optimizer=Adam(0.001), loss='mse')
    epsilon = 1.0
    for ep in range(n_episodes):
        q_all = qm.predict(X_dql, verbose=0)
        targets = q_all.copy()
        for i in range(len(X_dql)):
            action = np.random.randint(2) if np.random.rand() < epsilon else np.argmax(q_all[i])
            if action == y_dql[i]: reward = +1
            elif action == 0 and y_dql[i] == 1: reward = -2 if asymmetric else -1
            else: reward = -1
            targets[i][action] = reward
        qm.fit(X_dql, targets, epochs=5, batch_size=64, verbose=0)
        epsilon = max(0.05, epsilon * 0.85)
    preds = np.argmax(qm.predict(X_test, verbose=0), axis=1)
    return f1_score(y_test, preds, average='macro')

asym_f1s = []
sym_f1s  = []
for seed in seeds:
    a = run_reward(X_train, y_train, X_test, y_test, seed, asymmetric=True)
    s = run_reward(X_train, y_train, X_test, y_test, seed, asymmetric=False)
    asym_f1s.append(a)
    sym_f1s.append(s)
    print(f'Seed {seed} → Asymmetric: {a:.4f} | Symmetric: {s:.4f}')

asym_mean = np.mean(asym_f1s)
sym_mean  = np.mean(sym_f1s)
improvement = ((asym_mean - sym_mean) / sym_mean) * 100

print(f'\nSymmetric  : {sym_mean:.4f} +/- {np.std(sym_f1s):.4f}')
print(f'Asymmetric : {asym_mean:.4f} +/- {np.std(asym_f1s):.4f}')
print(f'Improvement: {improvement:.2f}%')
print('\n>>> USE THESE IN TABLE IV <<<')
```
